```python
from pyspark.sql import SparkSession
from graphframes import GraphFrame
from pyspark.sql.functions import col
from pyspark.sql.functions import col, desc, count


# ✅ Create Spark session (configured with DeltaCatalog in Docker)
spark = SparkSession.builder \
    .appName("FlightGraphAnalysis") \
    .getOrCreate()

# Optional: for cleaner notebook output
spark.conf.set("spark.sql.shuffle.partitions", spark._sc.defaultParallelism)
spark.conf.set("spark.sql.repl.eagerEval.enabled", True)
spark.conf.set("spark.sql.repl.eagerEval.truncate", 500)

print("✅ Spark session is ready")
```

```python
# Adjust path based on Docker volume mount: ../.data/ => /home/jovyan/input/
file_path = "/home/jovyan/input/2009.csv"

df = spark.read.csv(file_path, header=True, inferSchema=True)
df.printSchema()
df.show(5)
```

```python
# Show descriptive statistics for numerical columns
df.describe().show()

# Count unique airports in Origin and Destination
num_origin = df.select("Origin").distinct().count()
num_dest = df.select("Dest").distinct().count()
print(f"Unique Origin airports: {num_origin}")
print(f"Unique Destination airports: {num_dest}")
```

```python
# Vertices: All unique airport codes
vertices = df.select("Origin").union(df.select("Dest")) \
             .distinct().withColumnRenamed("Origin", "id")

# Edges: Flights from Origin to Dest
edges = df.select(col("Origin").alias("src"), col("Dest").alias("dst"))

# Create the GraphFrame
g = GraphFrame(vertices, edges)

# Sanity check
print("Number of vertices (airports):", g.vertices.count())
print("Number of edges (flights):", g.edges.count())
```

```python
## Query 1 – Degree and Triangle Statistics

"""
We analyze the flight graph by computing:
- **In-degree**: number of incoming flights to each airport.
- **Out-degree**: number of outgoing flights from each airport.
- **Total degree**: sum of in- and out-degrees.
- **Triangle count**: number of closed triplets A → B → C → A (computed using native Spark SQL joins).
"""
```

```python
# Count outgoing flights (from an airport)
out_deg = edges.groupBy("src").agg(count("*").alias("out_degree"))

# Count incoming flights (to an airport)
in_deg = edges.groupBy("dst").agg(count("*").alias("in_degree"))

# Join and calculate total degree
degree_df = in_deg.join(out_deg, in_deg["dst"] == out_deg["src"], "outer") \
                  .withColumn("airport", col("dst")) \
                  .drop("dst", "src") \
                  .fillna(0)

degree_df = degree_df.withColumn("total_degree", col("in_degree") + col("out_degree"))
degree_df.orderBy(desc("total_degree")).show(10)
```

```python
# Step 1: Create 2-hop paths: A -> B -> C
paths = edges.alias("e1").join(
    edges.alias("e2"),
    col("e1.dst") == col("e2.src")
).select(
    col("e1.src").alias("a"),
    col("e1.dst").alias("b"),
    col("e2.dst").alias("c")
)

# Step 2: Close the triangle: A -> B -> C -> A
triangles = paths.join(
    edges.alias("e3"),
    (col("e3.src") == col("c")) & (col("e3.dst") == col("a"))
)

# Step 3: Count triangles (each triangle appears 3 times)
triangle_count = triangles.count() // 3
print(f"Total number of triangles: {triangle_count}")
```
